```python
import numpy as np

print("NumPy Version:", np.__version__)
print("NumPy is ready!")
```

```python
# 1D Array
arr1 = np.array([10, 20, 30, 40, 50])

# 2D Array
arr2 = np.array([[1, 2, 3],
                 [4, 5, 6]])

# 3D Array
arr3 = np.array([[[1, 2], [3, 4]],
                 [[5, 6], [7, 8]]])

print("1D Array:\n", arr1)
print("\n2D Array:\n", arr2)
print("\n3D Array:\n", arr3)
```

```python
# Array Addition
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)

# Array Subtraction
print("Subtraction:", a - b)

# Array Multiplication
print("Multiplication:", a * b)

# Array Division
print("Division:", a / b)
```

```python
# Define two vectors
vector1 = np.array([1, 2, 3])
vector2 = np.array([4, 5, 6])

# Calculate Dot Product
result = np.dot(vector1, vector2)

print("Vector 1:", vector1)
print("Vector 2:", vector2)
print("Dot Product:", result)
```

```python
# Define two matrices
matrix1 = np.array([[1, 2],
                    [3, 4]])

matrix2 = np.array([[5, 6],
                    [7, 8]])

# Matrix Addition
print("Matrix Addition:\n", matrix1 + matrix2)

# Matrix Multiplication
print("\nMatrix Multiplication:\n", np.dot(matrix1, matrix2))

# Transpose of Matrix
print("\nTranspose of Matrix 1:\n", matrix1.T)
```

```python
# Define a 2x2 matrix
matrix = np.array([[4, 7],
                   [2, 6]])

# Calculate Determinant
det = np.linalg.det(matrix)

# Calculate Inverse
inverse = np.linalg.inv(matrix)

print("Original Matrix:\n", matrix)
print("\nDeterminant:", round(det, 2))
print("\nInverse Matrix:\n", inverse)
```

```python
# Create a 1D array
arr = np.array([1, 2, 3, 4, 5, 6])

# Reshape into a 2D array
reshaped = arr.reshape(2, 3)

# Flatten the 2D array into 1D
flattened = reshaped.flatten()

print("Original Array:", arr)
print("\nReshaped Array:\n", reshaped)
print("\nFlattened Array:", flattened)
```
